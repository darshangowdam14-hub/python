```python
def create_Risk_Client():
    print("Creating Risk API Client")
    return "Risk Client"

def get_risk(client,amount):
    if amount >= 50000:
        return "HIGH"
    elif amount >= 10000:
        return "MEDIUM"
    else:
        return "LOW"
```

```python
def process_Partitions(records):
    client = create_Risk_Client()

    for transection in records:
        transection_id = transection[0]
        client_id = transection[1]
        amount = transection[2]

        risk = get_risk(client,amount)

        yield(transection_id,
              client_id,
              amount,
              risk
        )

        
        
```

```python
result = rdd1.mapPartitions(process_Partitions)
result.collect()
```

```python
def salary_Calculation(employees):
    employee_id = employees[0]
    employee_name = employees[1]
    role = employees[2]
    salary = employees[3]
    rating = employees[4]

    bonus = get_bonus(salary,rating)
    final_salary = bonus+salary

    return(
            employee_id,
            employee_name,
            role,
            salary,
            rating,
            bonus,
            final_salary
        )
```

```python
# Q7 — Multiple Business Rules

transactions = [
    ("T001", "C101", 5000, "India", "UPI"),
    ("T002", "C102", 60000, "India", "CARD"),
    ("T003", "C103", 120000, "USA", "CARD"),
    ("T004", "C104", 8000, "UK", "CASH"),
    ("T005", "C105", 90000, "India", "UPI")
]

rdd = sc.parallelize(transactions)
```

```python
def calc_risk_amount(amount):
    if amount >= 100000:
        return 3
    elif amount >= 50000:
        return 2
    else:
        return 1

def calc_risk_country(country):
    if country != "India":
        return 2
    else:
        return 0

def calc_risk_Payment(pay_meth):
    if pay_meth == "CARD":
        return 1
    else:
        return 0 

def risk_calculation(risk_amount,risk_country,risk_pay):
    risk = risk_amount + risk_country + risk_pay
    if risk >= 5:
        return "HIGH"
    elif risk >= 3:
        return "MEDIUM"
    else:
        return "LOW"      
```

```python
def process_trans(transactions):
    tran_id,cust_id,amount,country,pay_meth = transactions

    risk_amount = calc_risk_amount(amount)
    risk_country = calc_risk_country(country)
    risk_pay = calc_risk_Payment(pay_meth)

    risk = risk_amount + risk_country + risk_pay
    risk_level = risk_calculation(risk_amount,risk_country,risk_pay)
  

    return (
        tran_id,
        cust_id,
        amount,
        country,
        pay_meth,
        risk,
        risk_level       
    )
```

```python
result = rdd.map(process_trans)
result.collect()
```

```python
#============================================================
# 8. mapPartitions() — Partition Level Data Quality Summary
# ============================================================

customers = [
    ("C101", "Anuj", 32, "anuj@gmail.com"),
    ("C102", "", 28, "rahul@gmail.com"),
    ("C103", "Priya", -5, "priya@gmail.com"),
    ("C104", "Amit", 45, ""),
    ("C105", "Neha", 25, "neha@gmail.com"),
    ("C106", "", 0, ""),
    ("C107", "Ravi", 40, "ravi@gmail.com"),
    ("C108", "Simran", 29, "simran@gmail.com")
]

rdd = sc.parallelize(customers, 4)
```

```python
def summarize_partitions(records):
    total_records = 0
    valid_records = 0
    invalid_records = 0

    for cust_id,name,age,email in records:
        total_records += 1

        if name != "" and age > 0 and email != "":
            valid_records += 1
        else:
            invalid_records += 1

    yield(
        total_records,valid_records,invalid_records
    )            

  
```

```python
result = rdd.mapPartitions(summarize_partitions)
result.collect()
    
```

```python
# ============================================================
# 9. mapPartitionsWithIndex() — Partition Distribution Analysis
# ============================================================

transactions = [
    ("T001", 5000),
    ("T002", 7000),
    ("T003", 9000),
    ("T004", 12000),
    ("T005", 15000),
    ("T006", 20000),
    ("T007", 25000),
    ("T008", 30000),
    ("T009", 35000),
    ("T010", 40000)
]

rdd = sc.parallelize(transactions, 4)
```

```python
def process_transformation(partition_id,records):
    total_amount = 0
    record_count = 0

    for tran_id,amount in records:
        record_count += 1
        total_amount += amount

    yield(partition_id,record_count,total_amount)  

result = rdd.mapPartitionsWithIndex(process_transformation)
result.collect()      
```
